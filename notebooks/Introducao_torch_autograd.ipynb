# Fundamentos de rede com pytorch

## Inicializando um modelo apartir do ResNet18 (modelo pré-treinado)

```python
import torch
from torchvision.models import resnet18, ResNet18_Weights
model = resnet18(weights=ResNet18_Weights.DEFAULT)
data = torch.rand(1, 3, 64, 64)
labels = torch.rand(1, 1000)
```

## Executando o forward

```python
prediction = model(data) # passagem para frente
```

## Executando o backward


Usamos a predição e o rótulo para calcular o erro (loss). A propagação para trás é iniciada ao chamar .backward() no tensor do erro. O Autograd calcula e armazena os gradientes de cada parâmetro no atributo .grad.

```python
loss = (prediction - labels).sum()
loss.backward() # backward pass
```

## Carregamos o optimizador

```python
optim = torch.optim.SGD(model.parameters(), lr=1e-2, momentum=0.9)
```

## Executamos o gradient descent

O backward preenche o .grad de cada parâmetro. O .step() lê esses .grad, combina com o learning rate e aí sim altera os pesos.

```python
optim.step() #gradient descent
```
